```python
# Problem #10 — Reverse a Number

# Write a Python program to reverse the digits of a number using a while loop.

# Example
```

```python
num = int(input())
nums = num
reverse = 0
while num:
    reverse = reverse * 10 + num % 10
    num = num // 10
if nums == reverse:
    print("this number is palindrome")
else:
    print("this number is not palindrome")    

```

```python
# 🎯 Problem 1 — Second Largest Digit

# # Write a Python program that takes an integer and finds the second largest digit.
```

```python
num = int(input())
largestnumber = 0
secondlargest = 0
while num:
    digit = num % 10
    if digit > largestnumber:
        secondlargest = largestnumber
        largestnumber = digit
    elif digit > secondlargest and digit < largestnumber:
        secondlargest = digit
        
    num = num // 10
print(largestnumber)    
print(secondlargest)        
```

```python
n = 5
for i in range(n):
    for j in range(2 * i + 1):
        if j <= i:
            print(j + 1,end=" ")
        else:  
            print(2 * i - j + 1,end=" ")
    print()    
```
