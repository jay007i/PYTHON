```python
num1 = int(input())
num2 = int(input())
num3 = int(input())
num4 = int(input())
num5 = int(input())

largest = num1
for i in (num2,num3,num4,num5):
    if largest > i :
        largest = i
print(largest)        
```

```python
# Problem #10 — Reverse a Number

# Write a Python program to reverse the digits of a number using a while loop.

# Example
```

```python
num = int(input())
nums = num
reverse = 0
while num:
    reverse = reverse * 10 + num % 10
    num = num // 10
if nums == reverse:
    print("this number is palindrome")
else:
    print("this number is not palindrome")    

```

```python
num = int(input())
Even = 0
Odd = 0

while num:
    if num % 2 == 0 :
        Even = Even + 1
    else :
        Odd = Odd + 1
    num = num // 10
print(f"Even : {Even} \nOdd : {Odd}")            
```

```python
# Next problem 🚀

# Problem: Count the sum of even digits and sum of odd digits separately.
```

```python
num = int(input())
sumofeven = 0
sumofodd = 0
while num:
    if num % 2 == 0:
        sumofeven = sumofeven + num % 10
    else:
        sumofodd = sumofodd + num % 10
    num = num // 10
print(f"sumofeven : {sumofeven} \n sumofodd : {sumofodd}")            
```
