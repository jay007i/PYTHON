```python
# 🟢 Problem 6 — Divisible by 5

# Write a Python program that takes an integer and checks whether it is divisible by 5.
```

```python
number = int(input())
if number % 5 == 0 and number % 3 == 0:
    print("this number is divisible by 5 and 3")
elif number % 5 == 0 and number % 3 != 0:
    print("this number is only divisible by 5")
elif number % 5 != 0 and number % 3 == 0:
    print("this number is only divisible by 3")
else:
    print("this number is not divisible by 5 and 3")    
```

```python
# 🟢 Problem 10 — Grade Calculator

# Write a program that takes marks from 0 to 100 and prints:

# 90–100 → A
# 75–89  → B
# 60–74  → C
# 40–59  → D
# 0–39   → F
```

```python
marks = int(input())

if 90 <= marks <= 100 :
    print("A")
elif 75 <= marks <= 89 :
    print("b")
elif 60 <= marks <= 74 :
    print("c")
elif 40 <= marks <= 59:
    print("D")   
elif 0 <= marks <= 39:
    print("Padho Yaa Maro")       
else :
    print("invalid number")          
```
