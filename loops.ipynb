```python
num1 = int(input())
num2 = int(input())
num3 = int(input())
num4 = int(input())
num5 = int(input())

positives = 0
negatives = 0
zeros = 0
for i in (num1,num2,num3,num4,num5):
    if i < 0:
        negatives = negatives + 1
    elif i > 0:
        positives = positives + 1
    else:
        zeros = zeros + 1

print(f"Positive Numbers : {positives} \nNegatives Numbers : {negatives} \nTotal Zeros : {zeros}")            
```
